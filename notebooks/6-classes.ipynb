# Classes
Classes are data structures that belong to the realm of OBJECT-ORIENTED PROGRAMMING (OOP).

A class represents an object (instance) and the operations that can be performed within the context of this object.

Values of a class are known as 'attributes' and functions within the class are 'methods'.

## 0. Intuition
Let us build intuition about Classes by thinking of a simple social network!

```python
import uuid

class MySocialNetwork:
    #This is called a constructor, meaning it 'constructs' the SocialNetwork object
    def __init__(self, name:str):
        #classes have attributes and methods:
        # attributes: variables of the class
        # methods: functions of the class

        #when we think of social networks (i.e., Instagram, TikTok,...), what could be an attribute of it?
        self.name = name #name of the network
        self.users = {} #the users of the social network
        self.posts = {} #the posts created by the users
        print(f'[LOG @ {self.name}] Created my social network!')
        #Note: 'self' represents 'this' in the context of the class. This means 'users' can be accessed from the object of MySocialNetwork

    #our simple social network requires some methods as well, like adding a user:
    def add_user(self, username:str, name:str, email:str):
        '''
        :param username: username of the user to add to the social network
        :param name: real name of the user
        :param email: email of the user
        :return: returns the user object
        '''

        if username not in self.users.keys():
            self.users[username] = User(self, username, name, email) #add user
            print(f'[LOG @ {self.name}] User {username} joined!')
            return self.users[username]
        else:
            print(f'[LOG @ {self.name}] User {username} already exists!')
            return None

    def remove_user(self, username:str):
        '''
        :param username:
        '''

        if username in self.users.keys():
            self.users.pop(username, None)
            print(f'[LOG @ {self.name}] User {username} removed!')
        else:
            print(f'[LOG @ {self.name}] User {username} does not exist!')

    def n_users(self):
        '''
        :return: the number of users in our social network
        '''

        return len(self.users)

    def n_posts(self):
        '''
        :return: the number of posts in our social network
        '''

        return len(self.posts)
```

A social network requires users, so we must create a class for the users

```python
class User:
    def __init__(self, network:MySocialNetwork, username:str, name:str, email:str):
        '''
        :param network: MySocialNetwork object
        :param username: the username of the user
        :param name: the real name of the user
        :param email: the email of the user
        '''

        self.network = network
        self.username = username
        self.name = name
        self.email = email
        self.following = {} #a structure to store the Users that this User follows
        self.followers = {} #a structure to store the Users that follow this User
        self.my_posts = {} #a structure to store the posts that the User makes
        self.n_followers = 0 #number of followers
        self.n_following = 0 #number of Users the user follows

    def follow(self, user_to_follow):
        '''
        :param user_to_follow: User object to follow
        '''

        if user_to_follow.username in self.network.users.keys():
            if user_to_follow.username in self.following.keys():
                print(f'[LOG @ {self.username}] User {user_to_follow.username} already followed!')
            elif user_to_follow.username == self.username:
                print(f'[LOG @ {self.username}] You cannot follow yourself!')
            else:
                self.following[user_to_follow.username] = user_to_follow
                self.n_following += 1 #increase the number of users the user follows
                user_to_follow.followers[self.username] = self
                user_to_follow.n_followers += 1 #increase the number of followers of the target user
                print(f'[LOG @ {self.username}] {self.username} started following {user_to_follow.username}!')
        else:
            print(f'[LOG @ {self.username}] User {user_to_follow.username} does not exist!')

    def unfollow(self, user_to_unfollow):
        '''
        :param user_to_unfollow: User object to follow
        '''

        if user_to_unfollow.username in self.network.users.keys():
            if user_to_unfollow.username not in self.following.keys():
                print(f'[LOG @ {self.username}] User {user_to_unfollow.username} is not followed!')
            elif user_to_follow.username == self.username:
                print(f'[LOG @ {self.username}] You cannot unfollow yourself!')
            else:
                self.following.pop(user_to_unfollow.username, None)
                self.n_following -= 1 #reduce the number of users the user follows
                user_to_unfollow.followers.pop(self.username, None)
                user_to_unfollow.n_followers -= 1 #reduce the number of followers of the target user
                print(f'[LOG @ {self.username}] {self.username} stopped following {user_to_unfollow.username}!')
        else:
            print(f'[LOG @ {self.username}] User {user_to_unfollow.username} does not exist!')

    def create_post(self, content):
        '''
        :param content: content of the post
        :return the ID of the post
        '''

        post_id = str(uuid.uuid4()) #random ID for the post
        current_post = Post(post_id, self.username, content) #create the object of the post
        self.my_posts[post_id] = current_post
        self.network.posts[post_id] = current_post #add the post to the global posts
        print(f'[New post from @{self.username}] {content}')
        
        return post_id

    def remove_post(self, post_id):
        '''
        :param post_id: ID of the post
        '''

        if post_id in self.my_posts.keys():
            self.my_posts.pop(post_id, None) #remove the post
            self.network.posts.pop(post_id, None) #remove the post from global posts
        else:
            print(f'[LOG @ {self.username}] Post {post_id} does not exist!')

    def like_post(self, post_id):
        '''
        :param post_id: ID of the post
        '''

        if post_id in self.network.posts.keys():
            self.network.posts[post_id].n_likes += 1 #add a like
            print(f'[LOG @ {self.username}] Liked post {post_id} from {self.network.posts[post_id].username}\ncontent: {self.network.posts[post_id].content}')
        else:
            print(f'[LOG @ {self.username}] Post {post_id} does not exist!')

    def remove_like_post(self, post_id):
        '''
        :param post_id: ID of the post
        '''

        if post_id in self.network.posts.keys():
            if self.network.posts[post_id].n_likes > 0:
                self.network.posts[post_id].n_likes -= 1 #remove a like
            else:
                print(f'[LOG @ {self.username}] Post {post_id} does not have enough likes to remove a like from it!')
        else:
            print(f'[LOG @ {self.username}] Post {post_id} does not exist!')
```

Users can also post, let's assume our social network is a message board (only accepts texts)

```python
class Post:
    def __init__(self, post_id:str, username, content:str):
        '''
        :param post_id: id generated of the post
        :param username: which user creates the post
        :param content: the content of the post
        '''

        self.post_id = post_id
        self.username = username
        self.content = content
        self.n_likes = 0 #number of likes
```

## 3. Following Users
Start following users with the "follow()" method from the User class
