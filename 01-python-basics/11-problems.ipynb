```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import gausspulse

def define_p():
  fs = 50e6
  fc = 3e6
  t_init = 0
  t_end = 1.5e-6
  noise_var = .03
  t = np.arange(t_init, t_end, 1/fs)
  samples = np.arange(len(t))
  p = np.roll(gausspulse(t-9e-7, fc=fc, bw=.8), 0)
  return t, p

t, p = define_p()

plt.figure(figsize=(8,2))
plt.stem(t, p)
plt.grid()
plt.xlabel('t [s]')
plt.ylabel('Amplitude')
plt.title('Array "p"')

def define_img():
  N = 100
  x = np.arange(N)
  y = np.arange(N)
  X, Y = np.meshgrid(x, y)
  img = X + Y*.2
  dist = np.sqrt((X-70)**2 + (Y-45)**2)
  mask = np.array((dist<10)+(dist > 30), dtype=bool)
  img[mask] = 10 * np.cos(.03 * X[mask] * Y[mask]) * 7*np.exp(-(X[mask]+10) * (Y[mask]+10)*1e-3)
  img[img>0] = img[img>0] - np.min(img[img>0])/2
  img[80, :] = img.max()/2
  return img

img = define_img()

plt.figure(figsize=(5,5))
plt.imshow(img)
plt.colorbar()
plt.title('Matrix"img"')
plt.xlabel('x')
plt.ylabel('y')
```
