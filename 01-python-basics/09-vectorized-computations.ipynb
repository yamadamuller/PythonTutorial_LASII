# Vectorized Computations
By now, you probably have a clue as to why NumPy is so important for scientific computing: vectorized computations. In short, vectorization is performing mathematical operations on an entire array with one command. This is possible because the content of the array is homogeneous, or, simply put, the elements of the array all belong to the same type. Since the interpreter "knows" that the content is homogeneous, with NumPy it is possible to avoid checking the data type of every element in a loop around the array, which has to be done for structures like lists. As a result, this optimized approach can significantly speed up your computations.

Let us try a computationally costly task, i.e., summing all integers from 0 to 900000:

```python
import numpy as np
import time

int_range = np.arange(0, 900001, 1) #"np.arange(start, stop, step)" arranges your series from within the half-open interval [start,stop)
#note that the interval DOES NOT include "stop"!
#another option would be using "np.linspace(start, stop, num)" which returns evenly spaced numbers over the specified 'num' interval
int_linspace = np.linspace(0,900000,900001)
```

```python
all(int_range==int_linspace) #check if the arrays match in all elements
```

```python
#First, let's sum the integers in the iterative way

int_sum = 0 #variable to store the iterative sum
t_init_iter = time.time() #variable to store when the algorithm started to process
for i in int_range:
    int_sum += i #increase the summ value
t_end_iter = time.time()-t_init_iter #registed how many seconds it took to compute the sum
print(f'Iterative sum = {int_sum} | computed in {t_end_iter}s')
```

```python
#Now, vectorized with numpy arrays

t_init_vec = time.time()
int_sum_vec = np.sum(int_range)
t_end_vec = time.time()-t_init_vec #registed how many seconds it took to compute the vectorized sum
print(f'Vectorized sum = {int_sum_vec} | computed in {t_end_vec}s')
```
