## 1. Functions
Another *very* important feature of programming in general (this is not exclusive to Python!) is FUNCTIONS.

Defined using the keyword "def", followed by the function name, the function parameters inside the parentheses, and a colon

They allow you to compute the same operation multiple times, subject to changes in the input arguments.

In practice, this means only one line of code each time you need to repeat the computation.

For example, if we wanted to create a function that computes the distance between two 2D-points (tuples), it would be something like:

```python
def euclid_distance(p1:tuple, p2:tuple):
    '''
    :param p1: point 1 (tuple by default)
    :param p2: point 2 (tuple by default)
    :return: the Euclidean distance between point 1 and 2
    '''

    return np.sqrt((p2[0]-p1[0])**2 + (p2[1]-p1[1])**2)

x1 = (2,-2)
x2 = (-2,6)
dist = euclid_distance(x1, x2)
print(f'dist(x1,x2) = {dist}')
```

Compare with the built-in distance function of the math package:

```python
dist_math = math.dist(x1, x2)
print(f'math.dist(x1,x2) = {dist}')
```

### 1.1 Lambda Functions
Sometimes, defining a function that takes one line with the "def" option can be unnecessary. In this case, there is the option of "Lambda" functions, or functions without a proper name. 

```python
dist_lambda = lambda p1, p2: np.sqrt((p2[0]-p1[0])**2 + (p2[1]-p1[1])**2) #distance between p1 and p2 as a Lambda function
print(f'dist_lambda(x1,x2) = {dist_lambda((2,-2),(-2,6))}')
```

Variables inside a function usually does not exist outside the function (exceptions if they are shadowed or the function is a method of a class, more on classes later!)

i.e., the Rosenbrock function is a very famous function in numerical optimization it is known as a benchmark function, given the difficulty to converge at its global minimum (1,1)

```python
def rosenbrock(x:tuple):
    '''
    :param x: a tuple with a 2D point to analyze its cost in the parameter space spanned by the Rosenbrock function
    :return: the value of the function at the evaluated input
    '''
    
    first_half = 100*(x[1]-x[0]**2)**2
    print(f'[inside the function] first_half = {first_half}')
    second_half = (1-x[0])**2
    print(f'[inside the function] second_half = {second_half}')
    return first_half + second_half

rosen_point = (-1.2,1)
cost = rosenbrock(rosen_point)
print(f'Rosenbrock{rosen_point} = {cost}')
```

## 2. "try" and "except"
Try to print the values computed in first_half and second_half outside the function

```python
try:
    print(f'[outside the function] first_half = {first_half}')
    print(f'[outside the function] second_half = {second_half}')
except:
    print(f'[outside the function] first_half and second_half do not exist outside rosenbrock()')
```

## 3. Default arguments
Functions can have default arguments when defined.

If the user does not provide them when calling the function, it defaults to the value provided in the definition.

```python
def power_of_n(x:float, n=2):
    '''
    :param x: number to evaluate its n-th power
    :param n: which n will be evaluated as power
    :return the n-th power of x
    '''
    
    return x**n

x3 = 2.
p_default = power_of_n(x3)
print(f'[power_of_n] default n = {p_default}')
p_argument = power_of_n(x3, n=3)
print(f'[power_of_n] "n=3": {p_argument}')
```

Functions in Python can return one or multiple values, but also nothing!

i.e., a function that prints all the characters in a name isn't required to return the input string

```python
print('Function with no return value:')
print()
def print_name(name:str):
    '''
    :param name: string with tour first name to iterate over and print its characters
    '''
    
    for i in name:
        print(i)
    print()

print_name('marina')
```
